```python
import numpy as np
a=np.array([1,2,3])
c=np.arange(0,10,2)            # range 
d=np.arange(1,11,2)
b=np.linspace(0,1,5)
print(a)
print(b)
print(c)
print(d)
print(np.linspace(1,5,11)) #divide the number by 6  default 50
print(np.linspace(0,10,5,endpoint=False))
```

1D,2D,3D array

```python
arr_1d=np.array([1,2,3])           #1D
arr_2d=np.array([[1,2,3],[4,5,6]])
arr_3d=np.array([[[1,2],[3,4],[5,6]]])
print(arr_1d,"\n")
print(arr_2d,"\n")
print(arr_3d)
```

special array

```python
z=np.zeros((2,3))
o=np.ones((2,3))
i=np.identity(3)
f=np.full((3,4),7)
print(f"zeros: {z}\n")
print(f"ones: {o}\n")
print(f"identity: {i}\n")
print(f"full: {f}\n")
```

array attributes

```python
arr=np.array([[1,2,3],[4,5,6]])
print(f"shape:{arr.shape}")
print(f"size:{arr.size}")
print(f"dimention:{arr.ndim}")
print(f"datatype:{arr.dtype}")
```

accessing array elements

```python
ar=np.array([1,2,3,4,5])
print(ar[0])
print(ar[-1])
print(ar[::2])
ar1=np.array([[1,2,3],[4,5,6],[7,9,10]])
print(ar1[0:3,1:3])
print(ar1[:,0:2])
```

array operations

```python
#arithmetic operation
a1=np.array([1,2,3])
b1=np.array([4,5,6])
print(a1+b1)
print(a1-b1)
print(a1*b1)
print(a1/b1)
```

```python
#broadcasting
arr2=np.array([[1,2,4],[5,6,3]])
b2=np.array([10,20,30])
print(arr2+b2)

c=np.array([[1],[2],[3]])
print(c)
d=np.array([10,20,30])
print(c+d)
```

```python
#scalar
arrr=np.array([1,2,3])
print(arrr**2)
print(arrr+10)
print(arrr*2)
print(arrr/2)
```
