```python
age=int(input("Enter age"))
if age<13:
    print("Child")
elif age<=19:
    print("Teenager")
elif age<=59 :
    print("Adult")
else:
    print("Senior")
```

```python
month=int(input("enter month number"))
if month==1:
    print("january")
elif month==2:
    print("february")
elif month==3:
    print("march")
elif month==4:
    print("April")
elif month==5:
    print("may")
elif month==6:
    print("june")
elif month==7:
    print("july")
elif month==8:
    print("August")
elif month==9:
    print("september")
elif month==10:
    print("october")
elif month==11:
    print("november")
elif month==12:
    print("december") 
else:
    print("invalid")
```

```python
temp=int(input("enter temperature:"))
if temp>35:
    print("very hot")
elif temp>=25:
    print("hot")
elif temp>=15:
    print("pleasant")
else:
    print("cold")
```

```python
marks=int(input("Enter marks"))
attendance=int(input("enter attendance"))
if marks>=40:
    if attendance>=75:
        print("eligible")
    else:
        print("not eligible ")
else:
    print("not eligible due to low marks ")
```
