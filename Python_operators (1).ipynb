# Arithmetic Operators

```python
#1)
length=12
width=5
c=2 * (length + width)
print("perimeter of a rectangle:",c)
d=length * width
print("area of a rectangle:",d)
```

```python
#2)
distance=150
time=3
b=distance/time
print("speed:",b)
```

```python
#3)
num=100
b=100**2
print("square:",b)
c=100**3
print("cube:",c)
```

```python
#4)
a=7
b=3
c=a**b
print("a**b:",c)
d=b**a
print("b**a:",d)
```

```python
#5)
marks_obtained=450
total_marks=500
s=(marks_obtained / total_marks) * 100
print("percentage:",s)
```

```python
#6)
hours=5
rate_per_hour=250
d=hours*rate_per_hour
print("total wages:",d)
```

```python
#8)
num=144
square_root=num**0.5
print("square root:",square_root)
```

# Logical Operator

```python
#17)
is_member=True
purchase_amount=600
discount_threshold=500
print("Discount eligibility:",is_member and purchase_amount >= discount_threshold)
```

```python
#18)
has_license=True
age=20
min_age=18
print("Driving eligibility:",has_license and age >= min_age)
```

```python
#19)
signal="red"
print("vehicle should stop:",signal=="red")
print("vehicle should not stop:",not(signal=="red"))
```

```python
#20)
day="Sunday"
is_open=False
print(is_open or day=="Sunday")
```

```python
#21)
temp=35
humidity=80
print("weather is uncomfortable:",temp>30 and humidity>70)
```

```python
#22)
is_admin=False
is_moderator=True
print("access rights:",is_admin or is_moderator)
```

```python
#23)
battery=15
is_charging=False
print("low battery warning:",battery and is_charging==False)
```

```python
#24)
subscription_active=True
trial_expired=False
print("Access:",subscription_active or trial_expired)
```
